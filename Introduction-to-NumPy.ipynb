# Introduction to NumPy

**NumPy** (Numerical Python) is the fundamental package for scientific computing in Python.

It provides:
- A powerful N-dimensional array object called `ndarray`
- Fast mathematical operations on arrays
- Tools for linear algebra, Fourier transforms, and random numbers

# Import `NumPy`

````python
import numpy as np
````

# Creating first array

The simplest way to create an array is: `np.array()`

````python
array1 = np.array([1, 2, 3 ,4 ,5])
print(array1)
````

# Type of `array`

````python
print(type(array1))
````

# Differences between `array` and `list`
### We can do some operations on arrays

````python
w1 = np.array([1, 2, 3, 4])
w2 = np.array([10, 20, 30, 40])

print("w1 + w2:", w1 + w2)
print("w1 - w2:", w1 - w2)
print("w1 * w2:", w1 * w2)
print("w1 / w2:", w1 / w2)
````

## Comparison Operations

Comparison operators return an array of **booleans** (`True` / `False`), element by element.

````python
a = np.array([1, 2, 3, 4, 5])

print("a > 3 :", a > 3)
print("a == 3:", a == 3)
print("a != 3:", a != 3)
print("a <= 2:", a <= 2)
````

# Array creation methods

### `np.zeros()`

````python
z1 = np.zeros(2)
print(f"z1:\n {z1}")

print()

z2 = np.zeros((2, 3))
print(f"z2:\n {z2}")


````

### `np.ones()`

````python
o1 = np.ones(2)
print(f"o1:\n {o1}")

print()

o2 = np.ones((2, 4, 3))
print(f"o2:\n {o2}")
````

### `np.arange(start, stop, num)`

Like Python's `range()`, but returns a `NumPy array`.

````python
a1 = np.arange(4)
a2 = np.arange(2, 10)
a3 = np.arange(2, 10, 2)

print(f"a1:\n {a1}")
print()
print(f"a2:\n {a2}")
print()
print(f"a3:\n {a3}")
````

### `np.linspace(start, stop, num)`

Returns **`num` evenly spaced points** between `start` and `stop`.


````python
l1 = np.linspace(2, 3, 10)
print(f"l1:\n {l1}")
````

# Linear Algebra with NumPy


### Vectors and Matrices

In NumPy:
- A **vector** is a 1D array
- A **matrix** is a 2D array



````python
v = np.array([1, 2, 3])  

m = np.array([[1, 2],
              [3, 4],
              [5, 6]]) 
             

print(f"v:\n {v}")
print(f"M:\n {m}")
print()
print(f"v.shape: {v.shape}")
print(f"M.shape: {m.shape}")
````

### Dot product
np.dot(a, b)

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(f"np.dot(a, b): {np.dot(a, b)}")
````

### Matrix Multiplication
For matrices `A` (shape `m×n`) and `B` (shape `n×p`), the product `A @ B` has shape `m×p`.

````python
m1 = np.array([[1, 2],
              [3, 4]])          

m2 = np.array([[5, 6],
              [7, 8]])          

print("m1 @ m2:\n", m1 @ m2)
````

## `*` vs `@` 

- `A * B` → element-wise multiplication
- `A @ B` → matrix multiplication

````python
m1 = np.array([[1, 2],
              [3, 4]])          

m2 = np.array([[5, 6],
              [7, 8]])          

print(f"m1 @ m2:\n {m1 @ m2}")
print()
print(f"m1 * m2:\n {m1 * m2}")
````

### Determinant
`np.linalg.det(A)`

````python
m = np.array([[1, 2],
              [3, 4]])

print(f"det(m): {np.linalg.det(m)}")
````

### Inverse
`np.linalg.inv(A)`


````python
m = np.array([[1, 2],
              [3, 4]])

m_inv = np.linalg.inv(m)
print(f"m_inv:\n {m_inv}")
````
